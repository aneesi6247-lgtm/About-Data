```python
# working with BeautifulSoup
from bs4 import BeautifulSoup

with open ("Data_Scraping/data1.html", "r") as f:
    data = f.read()
soup = BeautifulSoup(data, "lxml") # lxml is used to make this process more faster  # now the data is structured

```

```python
soup.find("h3")  # find the first occurance
data_rel = soup.find_all("h3") # find all the occurances
# the data is in the form of the list so we used for loop to get the data
all_cunt = []
for val in data_rel:
    count = val.get_text(strip = True),  # get_text() is used to get the content in a tag and strip to get rid of spaces
    # print(val.find_parent("div"))

    pop = val.find_next("div").select("span.country-population")[0].get_text(strip=True)
    
    # print(val.attrs)
    all_cunt.append([count,pop])
all_cunt
```

```python
import pandas as pd
data_frame =pd.DataFrame(all_cunt, columns =["countries", "population"])
data_frame
data_frame.to_csv("clean_data.csv")
```

```python
## Web_scraping exercise
# from quotes get all the qutoes and authors with the life tag
# first fetch the data

page_count = 1

while True:
    url_quotes = f"https://quotes.toscrape.com/page/{page_count}/"
    page_data = requests.get(url_quotes)

    soup = BeautifulSoup(page_data.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("no More valid pages available")
        break
    with open(f"web_scraping_assig_data/data{page_count}.html", "w") as f:
        f.write(page_data.text)
        print(f"The data is downloaded from the page{page_count}")
    page_count = page_count + 1
```

```python
# Get the useful information

life_quotes_with_authors = []
num = 1
while num < 11:
    with open (f"web_scraping_assig_data/data{num}.html", "r") as f:
        fetched = f.read()
        soup = BeautifulSoup(fetched, "lxml")
        all_quotes = soup.select("div.quote")
    for q in all_quotes:
        all_tags = []
        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())
        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes_with_authors.append([author, text])
    num = num + 1 
print(life_quotes_with_authors)

df = pd.DataFrame(life_quotes_with_authors, columns = ["Author", "Quote"])
df
df.to_csv("web_scraping_assig_data/clean_data.csv", index = False)
```
