```python
import json
def get_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
        return data
```

```python
First_data = get_data("04.03.2026.json")
print(First_data)
print(type(First_data))
```

```python
def clean_data(data):
    raw_data = {"one": 1, "two": 2, "three":3, "four": 4, "five":5}
    clean_data = []
    unique_user = set()
    # Remove the data inconsistency
    for user in data:
        user_rating = str(user["rating"]).strip().lower()
        if (user_rating in raw_data):
            user_rating = raw_data[user_rating]
        print(user_rating)
        user["rating"] = user_rating
    # handle the missing data
        raw_val = user.get("age")
        if raw_val == None:
            user["age"] = None
    # De-duplication of data
        if(user["name"].strip() in unique_user):
            continue
        unique_user.add(user["name"])
        clean_data.append(user)
    return clean_data
        
```

```python
cln = clean_data(First_data)
```

```python
# to get the meaningful information from the data
# 1. What is the average rating of the product

def avg_rat(data):
    rating = 0
    count = 0
    for val in data:
        ratings = float(val["rating"])
        rating += ratings
        count += 1
    average = rating/count
    print(f"The average rating of the product is {average}")
#2. What is the percentage of users who gave a poor rating(<3)

def Poor_ratPer(data):
    count = 0 
    for val in data:
        if float(val["rating"]) <3:
            count +=1
    percentage = (count / len(data))*100
    print(f"The Percentage of poor rating is {percentage}")
```

```python
avg_rat(cln)
Poor_ratPer(cln)
```

```python
# Write a program to get the recommendations for different users
def recomendations(data):
    recom = []
    for val in data:
        cur_rec = {}
        cur_rec["name"] = val["name"]
        if (float (val["rating"])) > 4 :
            cur_rec["brand"] = "Apple"
        else:
            cur_rec["brand"] = "samsung"
        recom.append(cur_rec)
    return recom
```

```python
recomendations(cln)
```
