```python
oscar_movies = [2004, 2006, 2008, 2010]
oscar_movies_reveneu = [34, 67, 12, 45]
non_oscar_movies_reveneu = [45, 78, 73, 98]
plt.plot(oscar_movies, oscar_movies_reveneu,"o--r", label = "Oscar Movie") # we can define the marker, line, and color with this formate
plt.plot(oscar_movies, non_oscar_movies_reveneu, color = "orange", marker = ">", linestyle = "-.", linewidth= "3", label = "non oscar Movies")

plt.xlabel("Years")
plt.ylabel("reveneu in MN")
plt.title("Oscal vs Non Oscar Movies Reveneu")
plt.legend()
```

```python
# Bar charts
oscar_movies = [2004, 2006, 2008, 2010]
oscar_movies_reveneu = [34, 67, 12, 45]
plt.bar(oscar_movies, oscar_movies_reveneu, color = "purple") # plt.bar() to get the bar graph

plt.xlabel("Years")
plt.ylabel("reveneu in MN")
plt.title("Oscal vs Non Oscar Movies Reveneu")
plt.ylim(0, 80)
for i in range(len(oscar_movies)):
    plt.text(oscar_movies[i], oscar_movies_reveneu [i] + .8, oscar_movies_reveneu[i], ha = "center") # .8 to put the value a little bit higher in the y-axis
    
```

```python
# to compare two sets of data
import numpy as np
oscar_movies = [2004, 2006, 2008, 2010]
oscar_movies_reveneu = [34, 67, 12, 45]
non_oscar_movies_reveneu = [45, 78, 73, 98]
x = np.arange(len(oscar_movies)) # it will create array acording to the number of years which are 4 [1,2,3,4]
width = 0.4

plt.bar(x-width/2, oscar_movies_reveneu, width, label = "oscar Movies") # shifting to left
plt.bar(x+width/2, non_oscar_movies_reveneu, width, label = "non Oscar movies") # shifting to right

plt.xlabel("Years")
plt.ylabel("reveneu in MN")
plt.title("Oscal vs Non Oscar Movies Reveneu")
plt.legend()
plt.ylim(0, 120)
plt.xticks(x, oscar_movies)  # to replace the x values to years in x-axis
```

```python
# scatter plts

plt.scatter(oscar_movies_reveneu, non_oscar_movies_reveneu, alpha =1, cmap = "OrRd", c = oscar_movies_reveneu) # alpha define for transperancy
# and cmap is to define which bar you gonna use from the online documententation, 
plt.colorbar(label = "Revenue")
## add annotations
for i in range(len(oscar_movies)):
    plt.annotate(oscar_movies[i], xy = (oscar_movies_reveneu[i], non_oscar_movies_reveneu[i]))
```

```python
# multiple plots
oscar_movies = [2004, 2006, 2008, 2010]
oscar_movies_reveneu = [34, 30, 12, 45]
oscar_movies_last_dacade = [2014, 2016, 2018, 2020]
oscar_movies_reveneu_last_decade = [80, 62, 98, 70]

plt.scatter(oscar_movies, oscar_movies_reveneu, color = "red")
plt.scatter(oscar_movies_last_dacade, oscar_movies_reveneu_last_decade, color= "black")
from matplotlib.ticker import ScalarFormatter

plt.title("movies in the last two decades")
plt.xlabel("Year")
plt.ylabel("Reveneu")
plt.tight_layout()

```

```python
# pie charts

movies_names= ["avatar", "avengers", "Titanic", "lucx"]
oscar_movies_reveneu = [34, 30, 12, 45]
explode= [0, 0.6, 0.2, 0]
plt.pie(oscar_movies_reveneu, labels = movies_names, autopct= "%1.1f%%", 
       wedgeprops= {
           "edgecolor" : "orange",
           "linewidth": "2",
           "linestyle" : "--",
           "fill" : True}, 
       explode = explode, startangle= 90) # autopact to show the percentage # fill false to remove all the colors # explode to get the value outside
plt.show()

# customization is piecharts

```
