```python
# to compare two sets of data
import numpy as np
oscar_movies = [2004, 2006, 2008, 2010]
oscar_movies_reveneu = [34, 67, 12, 45]
non_oscar_movies_reveneu = [45, 78, 73, 98]
x = np.arange(len(oscar_movies)) # it will create array acording to the number of years which are 4 [1,2,3,4]
width = 0.4

plt.bar(x-width/2, oscar_movies_reveneu, width, label = "oscar Movies") # shifting to left
plt.bar(x+width/2, non_oscar_movies_reveneu, width, label = "non Oscar movies") # shifting to right

plt.xlabel("Years")
plt.ylabel("reveneu in MN")
plt.title("Oscal vs Non Oscar Movies Reveneu")
plt.legend()
plt.ylim(0, 120)
plt.xticks(x, oscar_movies)  # to replace the x values to years in x-axis
```

```python
# Stack Plots
# Days (x-axis)
days = [1, 2, 3, 4, 5, 6, 7]

# Stock A (steady growth)
stock_A = [100, 105, 110, 108, 115, 120, 125]

# Stock B (volatile)
stock_B = [90, 95, 85, 100, 92, 110, 105]

# Stock C (decline then recovery)
stock_C = [120, 115, 110, 105, 108, 112, 118]

# Stock D (almost stable)
stock_D = [70, 72, 71, 73, 72, 74, 73]

# Stock E (strong growth)
stock_E = [50, 55, 60, 70, 80, 95, 110]

import matplotlib.pylab as plt
plt.stackplot(days, stock_A, stock_B, stock_C, stock_D, stock_E)
plt.xlabel("Day")
plt.ylabel("Stock Value")
plt.legend()
plt.title("Stock Values per Day")
plt.grid(True)


```

```python
# subplots -- to show multiple plots in the same figure
plt.subplot(2, 2, 1)
plt.plot(days, stock_A)
plt.title("First Plot")

plt.subplot(2, 2, 2)
plt.scatter(days, stock_B)
plt.title("Second Plot")

plt.subplot(2, 2, 3)
plt.boxplot(days, stock_C)
plt.title("third Plot")

plt.subplot(2, 2, 4)
plt.plot(days, stock_D)
plt.title("Fourt Plot")
```

```python
# Modern Matplotlib
# Object Oriented APIs
# More prfessional way which is used at the industry level

fig, ax = plt.subplots(2,2) # this return two values figure and axes

ax[0][0].plot(days, stock_A)
ax[0][0].set_title("First")
ax[0][0].set_xlabel("days")
ax[0][0].set_ylabel("value")


ax[0][1].plot(days, stock_B)
ax[0][1].set_title("second ")

ax[1][0].plot(days, stock_C)
ax[0][0].set_title("Third")

ax[1][1].plot(days, stock_D)
ax[0][0].set_title("Fourth")


fig.tight_layout()
```

```python


## Example Problem
# Compare the Temperature of the cities over a week using subplots and loops

# Cities
cities = ["Munich", "Berlin", "Hamburg"]

# Days
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

# Temperatures (each city has 7 values for 7 days)
temperatures = [
    [18, 20, 22, 21, 19, 17, 16],  # Munich
    [17, 19, 21, 20, 18, 16, 15],  # Berlin
    [16, 18, 19, 18, 17, 15, 14]   # Hamburg
]
city_num = 0
fig, ax = plt.subplots(1,3)
for i in range(1): # for row
    for j in range(3): # for column
        if city_num < len(cities):
            ax[j].plot(days, temperatures[city_num])
           # ax[j].set_title(f"The Temperature of {cities[city_num]} over the week")
            city_num = city_num + 1

fig.tight_layout()
fig.suptitle("Cities Temperature over the Week")
        
    
```

```python
# by chatgbt

import matplotlib.pyplot as plt

cities = ["Munich", "Berlin", "Hamburg"]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

temperatures = [
    [18, 20, 22, 21, 19, 17, 16],
    [17, 19, 21, 20, 18, 16, 15],
    [16, 18, 19, 18, 17, 15, 14]
]

for i in range(len(cities)):
    plt.plot(days, temperatures[i], marker='o', label=cities[i])

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Comparison of Cities Over a Week")

plt.legend()
plt.grid(True)
plt.tight_layout()

# plt.ylim(bottom=0)# optional (start from 0)

plt.show()
```
