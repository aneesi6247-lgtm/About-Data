```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# Problem 1
# Plot a of scores histogram•
#Drawa vertical line for the mean & try to interpret the results.
random = np.random.normal(70, 10, 200)
mean = np.mean(random)
plt.hist(random)
plt.axvline(x = mean, color = "red")
```

```python
#Problme# 03
tip = sns.load_dataset("tips")
sns.boxplot(
    data = tip,
    x = "total_bill", 
    y = "day",
    hue = "sex")
```

```python
# Problem # 04
sns.set_theme(style = "ticks")
sns.scatterplot(
    data = tip, 
    x = "tip",
    y = "total_bill", 
    hue = "sex",
    color = "red",
    style = "smoker")
plt.xlabel("Tips")
plt.ylabel("Total_bills")
plt.title("Bills vs Tip")
```

```python
# Problem # 05
table1 = tip.drop_duplicates()
table = table1.pivot_table( index = "day", columns = "time", values = "tip", aggfunc="mean")
table
sns.heatmap(
    data = table,
    cmap = "coolwarm",
    annot = True,
    
)
plt.tight_layout()
```
