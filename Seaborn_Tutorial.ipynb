```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Seaborn -- is a python data visualization libraray based on the Matplotlit. 

sns.set_theme()  # in seaborn first of all we set the theme, here we choosed the default one

load = sns.get_dataset_names()
load
```

```python
choose = sns.load_dataset("tips")
print(type(choose))
choose.head()
```

```python
sns.set_theme(style = "ticks") # to get the lines on each values of axis instad of plan axis
sns.relplot(
    data = choose,
    x = "total_bill", 
    y = "tip",
    hue = "sex",
    size ="size",
    style = "smoker") # to change the points style, size to change the size of the point according to the value, kind = "line" to change the type of graph to line
```

```python
planets = sns.load_dataset("planets")

sns.lineplot( 
 data = planets,
x = "distance",
y = "year",
hue = "year")
```

```python
flights = planets = sns.load_dataset("flights")

sns.lineplot( 
 data = flights,
x = "year",
y = "passengers") # this shaded line is called confidance interval (Ci) which is 95% # and its showing meean plus mins ci
```

```python
# categorical plots
#1. Bar plots
sns.barplot(
    data = choose,
    x = "day",
    y = "tip",
    hue = "sex")
```

```python
#2. Box plots
sns.boxplot(
    data = choose,
    x = "day",
    y = "tip",
    hue = "sex")
```

```python
# distribution plots 
#1- histo graphs
sns.histplot(
    data = choose,
    x = "tip",
    bins=10)
```

```python
# Matrix plots
#1 Heatmaps

pivot_flight = flights.pivot(index = "year", columns = "month", values = "passengers")
sns.heatmap(
    data = pivot_flight,
    cmap = "coolwarm",
    annot = True,
    fmt = "d")  # fmt = d is used to make the size of values small according to the box size
plt.title("Passengers Info")
plt.tight_layout()
```
