```python
import numpy as np
import time
array = np.array([1,4,12,321,421,123])
print(array)
print(type(array))
```

```python
size = 10000_000
# time takes a python list
py_list = list(range(size))
strt= time.time()
py_list = [x**2 for x in py_list]
end = time.time()
print(f"Python list takes around {end-strt}")

# times takes by an numpy array
num_arry = np.array([py_list])
start = time.time()
numf_arry = num_arry**2
ending = time.time()
print(f"NumPy list takes around {ending-start}")
```

```python
# memory
import sys
print(f"python list size is {sys.getsizeinfo(py_list) * len(py_list} in bytes")
print(f"Numpy array size is {sys.getsizeinfo() * len(py_list} in bytes")
```
