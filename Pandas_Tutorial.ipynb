```python
keg = pd.read_csv("globalAirQuality.csv")
```

```python
# Selecting data
# 1. select columns 
keg["city"]
keg[["city","aqi"]]

# select rows, we used iloc and loc for accesing rows data, iloc for accessing with index and loc for accessing with labels

keg.loc[0:4]  # in loc the final values of label is also print

keg.iloc[0:4] # but with iloc the final values of the index is not included 

# select rows and colomns

keg.loc[0:4, ["city", "pm10", "co", "temperature"]]
keg.iloc[0:5,2:5]

# select single scaler value at and iat

keg.at[19, "city"]
keg.iat[10, 6]

# its a good practice to create a copy of the data when your are traying to change something because here it will create a view, in which
# if we change anything it will change the values in the original dataset

cities= keg["city"].copy()
cities[0] = "lahore"
print(cities)

```

```python
keg
```

```python
# Filtering of data
keg[keg["aqi"] < 30]
keg[(keg["aqi"] < 30) & (keg["temperature"] <30)]

keg[keg["aqi"] < 30][["city", "country"]]

# Query Based Filtering:

keg.query("aqi>200")

var = 100
keg.query("pm25 > @var")  # @ is used to reference the python variables in the query
```

```python
# Handling the Missing values
empdata = pd.read_csv("raw_data.csv")
empdata.isnull()
empdata.isnull().sum()    # Return the number of the mising values in the columns
empdata.dropna()          # Drop all the rows which contain the missing values and only shows rows which contain all the values
empdata.dropna(axis=1)    # Drop all the columns which contain any missing values

clean_data = empdata.copy()
age_mean = clean_data["age"].mean()
clean_data["age"] = clean_data["age"].fillna(age_mean)
clean_data

empdata.ffill()  # Forward Fill
empdata.bfill()  # Backward Fill


# Handling the Duplicates

empdata["country"].duplicated()
empdata.drop_duplicates()
clean_data.drop_duplicates(inplace=True)   # inplace is used to directly make the changes in the original data set


clean_data
```
