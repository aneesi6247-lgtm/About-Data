```python
# cleaning the data

# Handling the Missing values
empdata = pd.read_csv("raw_data.csv")
empdata.isnull()
empdata.isnull().sum()    # Return the number of the mising values in the columns
empdata.dropna()          # Drop all the rows which contain the missing values and only shows rows which contain all the values
empdata.dropna(axis=1)    # Drop all the columns which contain any missing values

clean_data = empdata.copy()
age_mean = clean_data["age"].mean()
clean_data["age"] = clean_data["age"].fillna(age_mean)
clean_data

empdata.ffill()  # Forward Fill
empdata.bfill()  # Backward Fill


# Handling the Duplicates

empdata["country"].duplicated()
empdata.drop_duplicates()
clean_data.drop_duplicates(inplace=True)   # inplace is used to directly make the changes in the original data set


clean_data
```

```python

                # Transforming data (Feature Engineering)

#1-apply() to add a new feature column in the data with some logic
copy1 = empdata.copy()
copy1["tax"] = copy1["income"].apply(lambda x : "20%" if x < 55000 else"40%") # with lambda we directly write a function instead of
# writing it from the scratch. 
copy1

# 2-map() used to change the existing values to the new values
dic = { "Male" : "M", "Female" : "F", "Unknown" : "U"}
copy1["gender"] = copy1["gender"].map(dic)

# 3- assign() to assign new values in the new coloumn

copy1 = copy1.assign(new_income = copy1["income"] * 2)
copy1

# 4- replace(old, new) replace old values to the new ones

copy1["countr"] = copy1["country"].replace("USA", "US")
copy1 

#5- rename(column = {old:new} to change the coloumns of the data
copy1.columns = ["ID", "NAME", "AGE", "COUNTRY", "GENDER", "INCOME", "TAX", "NEW_INCOME","SOmething"]
copy1 = copy1.rename(columns = {"INCOME" : "SALARY"}) # for changing the name of the coloumns
# copy1 = copy1.rename(index = {1: "First"})   # for changing the vlues in the rows
copy1

# 6- sort_values() to sort out the values

sorted_val = copy1.sort_values("SALARY")
sorted_val

#copy1.sort_values("SALARY", ascending= False) # for decending order

# 7- sort_index()  # sort values based on index while reset index is used to the reset the index but still maintain the old index
sorted_val.reset_index()
sorted_val.reset_index(drop=True) # to delete the old index 
# 8- ranking

sorted_val["Ranking"] = sorted_val["NEW_INCOME"].rank(ascending = True, method = "min")  # in rank 1 is the loweest value to make it opposite ascending = true is used
sorted_val  # method is used to resolve the ties you can do it max, min, dance > to get the mean value
```

```python
# Melt and Pivot

#1- wide to long format

data = empdata.copy()
data.drop_duplicates()
melated_data = data.melt(
    id_vars = ["country", "name"],
    value_vars = ["age", "gender", "income"],
    var_name="Melted_data",
    value_name = "New Values")
    
    
melated_data
```

```python
#2. Pivot Function - to convert long data to wide data
# pivot() is used when we don't have duplicate data while the pivot_table is used to handle the duplicated data
original = melated_data.pivot_table(
    index= ["country", "name"],
    columns= ["Melted_data"],
    values= "New Values",
    aggfunc= "first").reset_index()  # moves the index back to columns and creates a new default integer index (0, 1, 2, …).
original
```
