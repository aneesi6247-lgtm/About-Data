```python
# creating arrays from scratch
arr = np.zeros((2,4)) # prefilled
print(arr)
arr1 = np.ones((4,)) # prefilled
print(arr1)
arr2 = np.full((3,4), 18) # prefillled with certain value
print (arr2)
arr3 = np.eye(2) # identity matrix
print(arr3)
arr4 = np.arange(0,11,3) # range elements
print(arr4) 
Array6= np.linspace(1, 500, 6) # Evenly spaced arrays
print(Array6)
```
